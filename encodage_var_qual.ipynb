```python
# variables catégorielles
vars_cat = ["Region", "Quality Score", "Income Group"]

# encodeur
oe = OneHotEncoder(
    drop="first",          # enlève une catégorie par variable (référence)
    handle_unknown="ignore"
)

# transformation
X_cat = oe.fit_transform(df[vars_cat])

# dataframe avec noms propres
X_cat = pd.DataFrame(
    X_cat,
    columns=oe.get_feature_names_out(vars_cat),
    index=df.index
)

# ajout au dataframe initial
df = pd.concat([df, X_cat], axis=1)

# suppression des colonnes catégorielles originales
df = df.drop(columns=vars_cat)

df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

oe_style = OneHotEncoder()
oe_results_region = oe_style.fit_transform(df[["Region"]])
pd.DataFrame(oe_results_region.toarray(), columns=oe_style.categories_).head()
```

```python
df = df.join(pd.DataFrame(oe_results_region.toarray(), columns=oe_style.categories_))
```

```python
oe_results_Q_score = oe_style.fit_transform(df[["Quality Score"]])
pd.DataFrame(oe_results_Q_score.toarray(), columns=oe_style.categories_).head()
```

```python
df = df.join(pd.DataFrame(oe_results_Q_score.toarray(), columns=oe_style.categories_))
```

```python
oe_results_I_group = oe_style.fit_transform(df[["Income Group"]])
pd.DataFrame(oe_results_I_group.toarray(), columns=oe_style.categories_).head()
```

```python
df = df.join(pd.DataFrame(oe_results_I_group.toarray(), columns=oe_style.categories_))
```
