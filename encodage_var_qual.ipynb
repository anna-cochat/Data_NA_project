```python
from sklearn.preprocessing import OneHotEncoder

oe_style = OneHotEncoder()
oe_results_region = oe_style.fit_transform(df[["Region"]])
pd.DataFrame(oe_results_region.toarray(), columns=oe_style.categories_).head()
```

```python
df = df.join(pd.DataFrame(oe_results_region.toarray(), columns=oe_style.categories_))
```

```python
oe_results_Q_score = oe_style.fit_transform(df[["Quality Score"]])
pd.DataFrame(oe_results_Q_score.toarray(), columns=oe_style.categories_).head()
```

```python
df = df.join(pd.DataFrame(oe_results_Q_score.toarray(), columns=oe_style.categories_))
```

```python
oe_results_I_group = oe_style.fit_transform(df[["Income Group"]])
pd.DataFrame(oe_results_I_group.toarray(), columns=oe_style.categories_).head()
```

```python
print("Nombre de lignes dans df :", len(df))
print("Nombre de lignes dans oe_results_I_group.toarray() :", oe_results_I_group.toarray().shape[0])
print("Nombre de catégories dans oe_style.categories_ :", len(oe_style.categories_))
print("Nombre de colonnes dans oe_results_I_group.toarray() :", oe_results_I_group.toarray().shape[1])
```

```python
#df = df.join(pd.DataFrame(oe_results_I_group.toarray(), columns=oe_style.categories_))
# Mettez à jour oe_style.categories_ pour qu'il ait 5 catégories
oe_style.categories_ = ['HI', 'LI', 'LM', 'UM', 'NaN']  # Remplacez par vos catégories réelles
df_oe_results = pd.DataFrame(oe_results_I_group.toarray(), columns=oe_style.categories_, index=df.index)

# Effectuez la jointure
df = df.join(df_oe_results)
```
